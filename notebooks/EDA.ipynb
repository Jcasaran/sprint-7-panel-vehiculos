# Análisis exploratorio de anuncios de vehículos
Objetivo: revisar calidad de datos y explorar las distribuciones de kilometraje y precio.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px

data_path = Path("../vehicles_us.csv")
cars = pd.read_csv(data_path)
cars.head()
```

```python
print("Filas y columnas:", cars.shape)
print("Valores ausentes por columna:")
display(cars.isna().sum().sort_values(ascending=False))
display(cars[["price", "odometer", "model_year"]].describe())
```

```python
odometer = cars.dropna(subset=["odometer"])
px.histogram(odometer, x="odometer", nbins=50, title="Distribución del kilometraje")
```

```python
pairs = cars.dropna(subset=["odometer", "price"])
pairs = pairs[(pairs["odometer"] >= 0) & (pairs["price"] > 0)]
px.scatter(pairs, x="odometer", y="price", opacity=0.35,
           hover_data=["model", "condition"], title="Precio frente a kilometraje")
```

## Conclusión
Los gráficos permiten ver la concentración de kilometrajes y la dispersión de precios. Hay valores ausentes y extremos; por ello, se eliminan los nulos de cada gráfico y solo se usan precios positivos en la comparación. La relación visual no implica causalidad.
